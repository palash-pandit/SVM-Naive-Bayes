2. What is the difference between Hard Margin and Soft Margin SVM4?

Hard margin: It demands strict linear separability with zero tolerance for misclassification or margin violations.

Soft margin: It allows the misscalssification of some datapoints for better generalization of decision boundary

3: What is the mathematical intuition behind SVM?


=>The core mathematical intuition behind a Support Vector Machine (SVM) is finding a decision boundary—called a hyperplane—that separates different classes with the largest possible margin.


In n dimensional space the equation of hyper plane is w^T*x+b=0,
where, w is the weight vectors that are perpendicular to the hyper plane.
x is the input feature vector.

b is the bias that shifts hyperplane from the origin.

4: What is the role of Lagrange Multipliers in SVM?

=>The kernel trick is a mathematical shortcut used in Support Vector Machines (SVM) to classify non-linearly separable data.


Instead of actually transforming or mapping data points into a higher-dimensional space (which is computationally expensive or impossible), the kernel trick computes the similarity (dot product) between data points as if they had been transformed.

11.What is the role of the Gamma parameter in RBF Kernel SVM?

=>In an RBF (Radial Basis Function) Kernel SVM, the gamma (γ) parameter controls how far the influence of a single training data point reaches.

Low gamma: Each data point has a wide area of influence. The decision boundary becomes smoother and simpler, which may cause underfitting.

High gamma: Each data point has a small, local area of influence. The decision boundary becomes more complex and may lead to overfitting.

24.Write a Python program to train an SVM Classifier with a Polynomial Kernel and visualize the decision
boundary:

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,mean_squared_error
import pandas as pd
from sklearn.datasets import make_circles
import numpy as np
import matplotlib.pyplot as plt


X,y=make_circles(n_samples=400,noise=0.2,random_state=1)

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.28,random_state=1)

cls=SVC(kernel="poly",degree=2,C=10,gamma='scale')
cls.fit(X_train,y_train)

y_pred=cls.predict(X_test)

acc=accuracy_score(y_test,y_pred)

print("accuracy score: ",acc)


x_min,x_max=X[:,0].min()-0.5,X[:,0].max()+0.5
y_min,y_max=X[:,1].min()-0.5,X[:,1].max()+0.5

xx,yy=np.meshgrid(
    np.linspace(x_min,x_max,500),
    np.linspace(y_min,y_max,500)
)

Z=cls.predict(np.c_[xx.ravel(),yy.ravel()])

Z=Z.reshape(xx.shape)

plt.contourf(xx,yy,Z,alpha=0.3)
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors="black")

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("SVM with Polynomial Kernel - Decision Boundary")
plt.show()
```

```python
from sklearn.svm import SVC
from sklearn.datasets import make_circles
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt

X,y=make_circles(n_samples=700,noise=0.2,random_state=1)

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.26,random_state=1)

C_val=[0.1,1,100]

plt.figure(figsize=(15,4))

for index,c in enumerate(C_val):

  cls=SVC(kernel="rbf",C=c)
  cls.fit(X_train,y_train)

  x_min,x_max=X[:,0].min()-0.5,X[:,1].max()+0.5
  y_min,y_max=X[:,1].min()-0.5,X[:,1].max()+0.5

  xx,yy=np.meshgrid(
      np.linspace(x_min,x_max,800),
      np.linspace(y_min,y_max,800)
  )


  Z=cls.predict(np.c_[xx.ravel(),yy.ravel()])
  Z=Z.reshape(xx.shape)

  plt.subplot(1,3,index+1)

  plt.contourf(xx,yy,Z,alpha=0.3)

  plt.scatter(X[:,0],X[:,1],c=y)

  plt.title(f"SVM Decision Boundary (C={c})")
  plt.xlabel("Feature 1")
  plt.ylabel("Feature 2")

  plt.show()
```

32.Write a Python program to train an SVM Classifier on an imbalanced dataset and apply class weighting and
check it improve accuracy

```python

from sklearn.svm import SVC
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

# Create imbalanced dataset
X, y = make_classification(
    n_samples=1000,
    n_features=10,
    n_classes=2,
    weights=[0.90, 0.10],
    random_state=1
)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1, stratify=y
)

# SVM without class weighting
cls1 = SVC(kernel='linear')

cls1.fit(X_train, y_train)

y_pred1 = cls1.predict(X_test)

acc1 = accuracy_score(y_test, y_pred1)

# SVM with class weighting
cls2 = SVC(kernel='linear', class_weight='balanced')

cls2.fit(X_train, y_train)

y_pred2 = cls2.predict(X_test)

acc2 = accuracy_score(y_test, y_pred2)

# Compare results
print("Accuracy without weighting:", acc1)
print("Accuracy with weighting:", acc2)

print("\nClassification Report with Weighting:")
print(classification_report(y_test, y_pred2))
```

36.Write a Python program to train an SVM Classifier using One-vs-Rest (OvR) and One-vs-One (OvO)
strategies on the Wine dataset and compare their accuracy

```python

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.multiclass import OneVsRestClassifier, OneVsOneClassifier
from sklearn.metrics import accuracy_score

# Load Wine dataset
data = load_wine()

X = data.data
y = data.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1, stratify=y
)

# Feature scaling
std = StandardScaler()

X_train = std.fit_transform(X_train)
X_test = std.transform(X_test)

# One-vs-Rest classifier
ovr = OneVsRestClassifier(SVC())

ovr.fit(X_train, y_train)

y_pred1 = ovr.predict(X_test)

acc1 = accuracy_score(y_test, y_pred1)

# One-vs-One classifier
ovo = OneVsOneClassifier(SVC())

ovo.fit(X_train, y_train)

y_pred2 = ovo.predict(X_test)

acc2 = accuracy_score(y_test, y_pred2)

# Compare accuracy
print("One-vs-Rest Accuracy:", acc1)
print("One-vs-One Accuracy:", acc2)
```

```python

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Load dataset
data = load_breast_cancer()

X = data.data
y = data.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

# Feature scaling
std = StandardScaler()

X_train = std.fit_transform(X_train)
X_test = std.transform(X_test)

# Linear Kernel
linear = SVC(kernel='linear')

linear.fit(X_train, y_train)

y_pred1 = linear.predict(X_test)

acc1 = accuracy_score(y_test, y_pred1)

# Polynomial Kernel
poly = SVC(kernel='poly')

poly.fit(X_train, y_train)

y_pred2 = poly.predict(X_test)

acc2 = accuracy_score(y_test, y_pred2)

# RBF Kernel
rbf = SVC(kernel='rbf')

rbf.fit(X_train, y_train)

y_pred3 = rbf.predict(X_test)

acc3 = accuracy_score(y_test, y_pred3)

# Compare accuracy
print("Linear Kernel Accuracy:", acc1)
print("Polynomial Kernel Accuracy:", acc2)
print("RBF Kernel Accuracy:", acc3)
```

38.Write a Python program to train an SVM Classifier using Stratified K-Fold Cross-Validation and compute the
average accuracy=

```python

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.svm import SVC

# Load dataset
data = load_breast_cancer()

X = data.data
y = data.target

# Create SVM model with feature scaling
cls = make_pipeline(
    StandardScaler(),
    SVC(kernel='rbf')
)

# Apply Stratified K-Fold
skf = StratifiedKFold(
    n_splits=5, shuffle=True, random_state=1
)

# Calculate accuracy for each fold
scores = cross_val_score(
    cls, X, y, cv=skf, scoring='accuracy'
)

# Display results
print("Accuracy of each fold:", scores)

# Calculate average accuracy
print("Average Accuracy:", scores.mean())
```

 41.Write a Python program to train an SVM Classifier and evaluate its performance using Precision, Recall, and
F1-Score instead of accuracy.

```python

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.feature_selection import RFE
from sklearn.metrics import accuracy_score

# Load dataset
data = load_breast_cancer()

X = data.data
y = data.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

# Feature scaling
std = StandardScaler()

X_train = std.fit_transform(X_train)
X_test = std.transform(X_test)

# SVM without feature selection
cls1 = SVC(kernel='linear')

cls1.fit(X_train, y_train)

y_pred1 = cls1.predict(X_test)

acc1 = accuracy_score(y_test, y_pred1)

# Apply RFE
model = SVC(kernel='linear')

rfe = RFE(estimator=model, n_features_to_select=10)

X_train_selected = rfe.fit_transform(X_train, y_train)
X_test_selected = rfe.transform(X_test)

# SVM with RFE
cls2 = SVC(kernel='linear')

cls2.fit(X_train_selected, y_train)

y_pred2 = cls2.predict(X_test_selected)

acc2 = accuracy_score(y_test, y_pred2)

# Compare accuracy
print("Accuracy without RFE:", acc1)
print("Accuracy with RFE:", acc2)

# Display selected features
print("Selected Features:", rfe.get_support().sum())
```

42. Write a Python program to train a Naïve Bayes Classifier and evaluate its performance using Log Loss
(Cross-Entropy Loss)

```python

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import log_loss, accuracy_score

# Load dataset
data = load_breast_cancer()

X = data.data
y = data.target

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

# Create Gaussian Naive Bayes model
cls = GaussianNB()

# Train model
cls.fit(X_train, y_train)

# Make predictions
y_pred = cls.predict(X_test)

# Predict probabilities
y_prob = cls.predict_proba(X_test)

# Calculate accuracy and log loss
acc = accuracy_score(y_test, y_pred)
loss = log_loss(y_test, y_prob)

# Display results
print("Accuracy:", acc)
print("Log Loss:", loss)
```

43.Write a Python program to train an SVM Classifier and visualize the Confusion Matrix using seaborn

```python

# Import required libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
data = load_breast_cancer()
X = data.data
y = data.target

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1
)

# Apply feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Create and train the SVM model
model = SVC(kernel='rbf')
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate accuracy and confusion matrix
accuracy = accuracy_score(y_test, y_pred)
cm = confusion_matrix(y_test, y_pred)

print("Accuracy:", accuracy)
print("Confusion Matrix:")
print(cm)

# Visualize the confusion matrix
sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',

)

plt.xlabel("Predicted Values")
plt.ylabel("Actual Values")
plt.title("SVM Confusion Matrix")
plt.show()
```

45. Write a Python program to train a Naïve Bayes classifier and evaluate its performance using the ROC-AUC

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score,roc_auc_score,roc_curve
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_breast_cancer
import matplotlib.pyplot as plt

data = load_breast_cancer()
X = data.data
y = data.target

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1, stratify=y
)

# Create and train the Naive Bayes model
model = GaussianNB()
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

y_prob=model.predict_proba(X_test)[:,1]

accuracy = accuracy_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_prob)

print("Accuracy:", accuracy)
print("ROC-AUC Score:", auc)

fpr,tpr,thersholds=roc_curve(y_test,y_prob)
plt.plot(fpr, tpr)
plt.plot([0, 1], [0, 1], linestyle="--", label="Random Classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - Naive Bayes")
plt.legend()
plt.show()
```

46. Write a Python program to train an SVM Classifier and visualize the Precision-Recall Curve.

```python

# Import required libraries
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import (
    precision_recall_curve,
    average_precision_score,
    accuracy_score
)
import matplotlib.pyplot as plt

# Load the dataset
data = load_breast_cancer()
X = data.data
y = data.target

# Split the dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=1, stratify=y
)

# Apply feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Create and train SVM classifier
model = SVC(kernel='rbf')
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Get decision scores for Precision-Recall curve
y_scores = model.decision_function(X_test)

# Calculate Precision and Recall
precision, recall, thresholds = precision_recall_curve(
    y_test, y_scores
)

# Calculate Average Precision
ap = average_precision_score(y_test, y_scores)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("Average Precision:", ap)

# Visualize Precision-Recall Curve
plt.plot(recall, precision, label="SVM (AP = %.2f)" % ap)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve - SVM")
plt.legend()
plt.grid()
plt.show()
```
